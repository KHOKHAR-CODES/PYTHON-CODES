```python
# ==========================================
# 1. INPUT & TYPE CONVERSIONS
# ==========================================
print("=== RAG Document Ingestion System ===\n")

# String inputs (default type of input() is str)
doc_id = input("Enter Document ID (e.g., DOC001): ")
title = input("Enter Document Title: ")
text_content = input("Enter Document Text Content: ")

# Numeric inputs with Type Conversion
total_pages = int(input("Enter total page count: "))
relevance_score = float(input("Enter initial relevance score (0.0 to 1.0): "))

# Boolean conversion (evaluates to True if user enters 'y' or 'yes')
is_confidential_input = input("Is this document confidential? (y/n): ").strip().lower()
is_confidential = is_confidential_input in ['y', 'yes']

# List creation via string splitting
raw_tags = input("Enter comma-separated tags (e.g., python, rag, ai): ")
tags = [tag.strip() for tag in raw_tags.split(",")]


# ==========================================
# 2. COMPUTED METADATA (using len)
# ==========================================
char_count = len(text_content)
word_count = len(text_content.split())
tag_count = len(tags)


# ==========================================
# 3. RAG METADATA DICTIONARY (Data Types)
# ==========================================
# Storing all attributes in a Dictionary—the standard structure for RAG payloads
document_record = {
    "doc_id": doc_id,                # String
    "title": title,                  # String
    "content": text_content,          # String
    "metadata": {
        "total_pages": total_pages,  # Integer
        "relevance": relevance_score,# Float
        "confidential": is_confidential, # Boolean
        "tags": tags,                # List of Strings
        "metrics": (char_count, word_count) # Tuple of (char_count, word_count)
    }
}


# ==========================================
# 4. DISPLAY & TYPE INSPECTION (using print & type)
# ==========================================
print("\n" + "="*40)
print("     DOCUMENT METADATA SUMMARY     ")
print("="*40)

print(f"ID: {document_record['doc_id']} | Type: {type(document_record['doc_id'])}")
print(f"Title: {document_record['title']} | Type: {type(document_record['title'])}")
print(f"Pages: {document_record['metadata']['total_pages']} | Type: {type(document_record['metadata']['total_pages'])}")
print(f"Relevance Score: {document_record['metadata']['relevance']} | Type: {type(document_record['metadata']['relevance'])}")
print(f"Confidential: {document_record['metadata']['confidential']} | Type: {type(document_record['metadata']['confidential'])}")
print(f"Tags ({tag_count}): {document_record['metadata']['tags']} | Type: {type(document_record['metadata']['tags'])}")
print(f"Metrics (Chars, Words): {document_record['metadata']['metrics']} | Type: {type(document_record['metadata']['metrics'])}")

print("\n--- Full Payload ready for Vector Database ---\n")
print(document_record)
```

```python
import curses
import random
import time

def main(stdscr):
    # Hide the mouse cursor and clear screen
    curses.curs_set(0)
    stdscr.nodelay(True)  # Don't block waiting for user input
    stdscr.timeout(100)   # Refresh screen every 100ms (game speed)

    # Get screen dimensions
    height, width = stdscr.getmaxyx()
    
    # Define track boundary dimensions
    track_left = width // 2 - 8
    track_right = width // 2 + 8
    
    # Player starting position (centered on track near bottom)
    car_x = width // 2
    car_y = height - 4
    
    # Game variables
    obstacles = []  # List of [y, x] pairs
    score = 0
    game_over = False

    while not game_over:
        # 1. HANDLE USER INPUT
        key = stdscr.getch()
        if key == curses.KEY_LEFT and car_x > track_left + 1:
            car_x -= 1
        elif key == curses.KEY_RIGHT and car_x < track_right - 1:
            car_x += 1
        elif key == ord('q'):
            break  # Quit game

        # 2. GENERATE OBSTACLES
        # 20% chance to spawn a new obstacle at the top each tick
        if random.random() < 0.2:
            obs_x = random.randint(track_left + 1, track_right - 1)
            obstacles.append([1, obs_x])

        # 3. UPDATE OBSTACLE POSITIONS
        new_obstacles = []
        for obs in obstacles:
            obs[0] += 1  # Move obstacle down by 1 line
            
            # Check for collision with player car
            if obs[0] == car_y and obs[1] == car_x:
                game_over = True
            
            # Keep obstacles on screen, score point when they pass bottom
            if obs[0] < height - 1:
                new_obstacles.append(obs)
            else:
                score += 10
                
        obstacles = new_obstacles

        # 4. DRAW EVERYTHING
        stdscr.clear()

        # Draw track boundaries
        for y in range(1, height - 1):
            stdscr.addch(y, track_left, '|')
            stdscr.addch(y, track_right, '|')

        # Draw obstacles
        for obs in obstacles:
            stdscr.addch(obs[0], obs[1], 'X')

        # Draw player car
        stdscr.addch(car_y, car_x, '▲')

        # Draw score HUD
        stdscr.addstr(0, 2, f"Score: {score}  |  Controls: ← / → Arrow Keys (Q to quit)")

        # Refresh display
        stdscr.refresh()

    # 5. GAME OVER SCREEN
    stdscr.nodelay(False)
    stdscr.clear()
    msg = f"GAME OVER! Final Score: {score}"
    sub_msg = "Press any key to exit..."
    stdscr.addstr(height // 2, (width - len(msg)) // 2, msg)
    stdscr.addstr(height // 2 + 1, (width - len(sub_msg)) // 2, sub_msg)
    stdscr.refresh()
    stdscr.getch()

# Run the curses wrapper
if __name__ == '__main__':
    curses.wrapper(main)
```
